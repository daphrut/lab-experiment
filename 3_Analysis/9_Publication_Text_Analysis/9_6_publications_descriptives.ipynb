# 9.6 Publication descriptives

This notebook prints some stats re publication data for the paper.

1. The number of labgroupids that consented to data merging.
2. The number of unique publications matched to a consenting researcher.
3. The number of labgroupids with at least one matched publication.

```python
# Set up
import pandas as pd
import sys
from pathlib import Path

CODE_ROOT = Path.cwd().parents[1]
sys.path.append(str(CODE_ROOT))
import config
```

```python
# Load data
labs = pd.read_csv(
    config.CLEAN_DATA / "final_dataset.csv",
    keep_default_na=False, # Keep "None" as a string, not NaN
    na_values=[""] # Only treat empty strings as NaN
)

publications = pd.read_csv(
    config.PUBLICATON_DATA /
    "2_Processed" /
    "publications_matched.csv"
)
```

## (1) Labgroupids that consented

```python
# Consenting to merge data
consented = labs["consent_data_merge"] == "Yes I consent to this data collection and merging"
n_consenting_labs = labs.loc[consented, "labgroupid"].nunique()

print(f"Labgroupids that consented to data merging: {n_consenting_labs:,}")
```

## (2) Unique publications

```python
# publications_matched.csv has one row per publication, with all of its matched
# labgroupids in one ";"-separated cell, so a publication matched to several
# consenting researchers is not repeated. We count distinct "relation" (the ZORA
# publication ID) to also guard against the same publication appearing in two rows.
n_unique_pubs = publications["relation"].nunique()

print(f"Unique publications matched to a consenting researcher: {n_unique_pubs:,}")
```

## (3) Unique labgroupids with at least one publication

```python
# Print the number of labgroupids that are matched to at least one publication 
# Note matched_labgroupids is e.g. "123" or "123;456"
matched_labgroupids = publications["matched_labgroupids"].str.split(";").explode().unique()
n_matched_labgroupids = len(matched_labgroupids)
print(f"Labgroupids matched to at least one publication: {n_matched_labgroupids:,}")
```
