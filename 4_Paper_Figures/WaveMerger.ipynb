```python
import pandas as pd
import numpy as np
import sys
import re
import importlib
from pathlib import Path
CODE_ROOT = Path.cwd().parents[0]
sys.path.append(str(CODE_ROOT))
import configW2 as config
import os
import statsmodels.formula.api as smf
import pyfixest as pf
```

```python
## Load data from wave 1
df_1 = pd.read_csv(
    config.DATA_ROOT / "17_Clean_Data/final_dataset.csv", # Wave 1
    keep_default_na=False, # Keep "None" as a string, not NaN
    na_values=[""] # Only treat empty strings as NaN
)

## Load data from wave 2
df_2 = pd.read_csv(
    config.CLEAN_DATA / "final_dataset.csv", 
    keep_default_na=False, # Keep "None" as a string, not NaN
    na_values=[""] # Only treat empty strings as NaN
)
```

```python
## Get columns that do not exactly match
col_1 = df_1.columns
col_2 = df_2.columns
extra_1 = col_1.difference(col_2) ## What is in 1, but not 2
extra_2 = col_2.difference(col_1) ## What is in 2, but not 1

## Missing columns do not appear in analysis, so all of them can be dropped
df_1 = df_1.drop(extra_1, axis=1)
df_2 = df_2.drop(extra_2, axis=1)
df = pd.concat([df_1, df_2], ignore_index=True)
```

```python
## Save data
df.to_csv(config.CLEAN_DATA / "merged_wave_final_dataset.csv", index=False)
```
